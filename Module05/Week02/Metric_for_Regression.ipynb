# **ME**

```python
actual = [3, 5, 2, 8, 7]
predicted1 = [2.5, 5.3, 1.9, 7.8, 7.5]
predicted2 = [3.5, 5.3, 2.1, 8.2, 7.5]
predicted3 = [2.5, 4.7, 1.9, 7.8, 6.5]
```

```python
def me(actual, predicted):
    actual, predicted = np.array(actual), np.array(predicted)
    errors = actual - predicted
    return np.mean(errors)
```

```python
me(actual, predicted1), me(actual, predicted2), me(actual, predicted3)
```

# **MAPE**

```python
actual = [3, 5, 2, 8, 7]
predicted1 = [2, 2, 2, 2, 2]
predicted2 = [5, 5, 5, 5, 5]
predicted3 = [7, 7, 7, 7, 7]
```

```python
def mape(actual, predicted):
    actual, predicted = np.array(actual), np.array(predicted)
    percentage_errors = np.abs((actual - predicted) / actual) * 100
    return np.mean(percentage_errors)
```

```python
mape(actual, predicted1), mape(actual, predicted2), mape(actual, predicted3)
```

```python
from sklearn.metrics import mean_absolute_percentage_error
```

```python
mean_absolute_percentage_error(actual, predicted1), mean_absolute_percentage_error(actual, predicted2), mean_absolute_percentage_error(actual, predicted3)
```

# **sMAPE**

```python
actual = np.array([3, 5, 2, 8, 7])
predicted1 = np.array([2, 2, 2, 2, 2])
predicted2 = np.array([5, 5, 5, 5, 5])
predicted3 = np.array([7, 7, 7, 7, 7])
```

```python
def smape(actual, predicted):
    actual, predicted = np.array(actual), np.array(predicted)
    numerator = np.abs(actual - predicted)
    denominator = np.abs(actual) + np.abs(predicted)
    smape_values = numerator / denominator
    return np.mean(smape_values) * 200
```

```python
smape(actual, predicted1), smape(actual, predicted2), smape(actual, predicted3)
```

```python
from sktime.performance_metrics.forecasting import MeanAbsolutePercentageError
```

```python
MeanAbsolutePercentageError(symmetric=True)(actual, predicted1), MeanAbsolutePercentageError(symmetric=True)(actual, predicted2), MeanAbsolutePercentageError(symmetric=True)(actual, predicted3)
```

# **MRAE**

```python
actual = np.array([3, 5, 2, 8, 7], dtype=float)
# Random-walk naive forecast (shifted actual values)
naive = np.roll(actual, 1)  # Shift the actual values by 1
naive[0] = np.nan  # First value is undefined for random-walk method
predicted1 = np.array([3.5, 3, 5, 2, 8])
predicted2 = np.array([3.5, 5.3, 2.1, 8.2, 7.5])
predicted3 = np.array([3.5, 2, 6, 1, 9])
```

```python
def mrae(actual, predicted, naive):
    actual, predicted, naive = np.array(actual), np.array(predicted), np.array(naive)
    rae = np.abs(actual - predicted) / np.abs(actual - naive)
    return np.mean(rae)
```

```python
mrae(actual[1:], predicted1[1:], naive[1:]), mrae(actual[1:], predicted2[1:], naive[1:]), mrae(actual[1:], predicted3[1:], naive[1:])
```

```python
from sktime.performance_metrics.forecasting import MeanRelativeAbsoluteError
```

```python
MeanRelativeAbsoluteError()(actual[1:], predicted1[1:], y_pred_benchmark=naive[1:]), MeanRelativeAbsoluteError()(actual[1:], predicted2[1:], y_pred_benchmark=naive[1:]), MeanRelativeAbsoluteError()(actual[1:], predicted3[1:], y_pred_benchmark=naive[1:])
```

# **GMRAE**

```python
actual = np.array([3, 5, 2, 8, 7], dtype=float)
# Random-walk naive forecast (shifted actual values)
naive = np.roll(actual, 1)  # Shift the actual values by 1
naive[0] = np.nan  # First value is undefined for random-walk method
predicted1 = np.array([3.5, 3, 5, 2, 8])
predicted2 = np.array([3.5, 5.3, 2.1, 8.2, 7.5])
predicted3 = np.array([3.5, 2, 6, 1, 9])
```

```python
def gmrae(actual, predicted, naive):
    return np.prod(np.abs((actual - predicted) / (actual - naive))) ** (1 / len(actual))
```

```python
gmrae(actual[1:], predicted1[1:], naive[1:]), gmrae(actual[1:], predicted2[1:], naive[1:]), gmrae(actual[1:], predicted3[1:], naive[1:])
```

```python
from sktime.performance_metrics.forecasting import GeometricMeanRelativeAbsoluteError
```

```python
GeometricMeanRelativeAbsoluteError()(actual[1:], predicted1[1:], y_pred_benchmark=naive[1:]), GeometricMeanRelativeAbsoluteError()(actual[1:], predicted2[1:], y_pred_benchmark=naive[1:]), GeometricMeanRelativeAbsoluteError()(actual[1:], predicted3[1:], y_pred_benchmark=naive[1:])
```

# **RGRMSE**

```python
actual = np.array([3, 5, 2, 8, 7], dtype=float)
# Random-walk naive forecast (shifted actual values)
naive = np.roll(actual, 1)  # Shift the actual values by 1
naive[0] = np.nan  # First value is undefined for random-walk method
predicted1 = np.array([3.5, 3, 5, 2, 8])
predicted2 = np.array([3.5, 5.3, 2.1, 8.2, 7.5])
predicted3 = np.array([3.5, 2, 6, 1, 9])
```

```python
def rgrmse(actual, predicted, naive):
    actual, predicted, naive = np.array(actual), np.array(predicted), np.array(naive)
    relative_squared_error = (np.abs(actual - predicted) / np.abs(actual - naive)) ** 2
    product_rse = np.prod(relative_squared_error)
    rgrmse_value = product_rse ** (1 / (2 * len(actual)))
    return rgrmse_value
```

```python
rgrmse(actual[1:], predicted1[1:], naive[1:]), rgrmse(actual[1:], predicted2[1:], naive[1:]), rgrmse(actual[1:], predicted3[1:], naive[1:])
```

```python
from sktime.performance_metrics.forecasting import GeometricMeanRelativeSquaredError
```

```python
GeometricMeanRelativeSquaredError(square_root=True)(actual[1:], predicted1[1:], y_pred_benchmark=naive[1:]), GeometricMeanRelativeSquaredError(square_root=True)(actual[1:], predicted2[1:], y_pred_benchmark=naive[1:]), GeometricMeanRelativeSquaredError(square_root=True)(actual[1:], predicted3[1:], y_pred_benchmark=naive[1:])
```
