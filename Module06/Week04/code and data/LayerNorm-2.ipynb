## pytorch

```python
import torch
import numpy as np

 # Image Example
N, C, H, W = 1, 2, 2, 2
data =  np.zeros((N, C, H, W))
data[0,0,:,:] = np.array([[8, 6],
                          [2, 4]])
data[0,1,:,:] = np.array([[0, 2],
                          [1, 5]])

data = data.reshape(N, C, H, W)
print(data.shape)
print(data[0,0,:,:])
print(data[0,1,:,:])
```

```python
input = torch.tensor(data,dtype=torch.float)
layer_norm = torch.nn.LayerNorm([C, H, W])
output = layer_norm(input)
print(output)
```

## self-implementation

```python
import numpy as np

N, C, H, W = 1, 2, 2, 2
data =  np.zeros((N, C, H, W))
data[0,0,:,:] = np.array([[8, 6],
                          [2, 4]])
data[0,1,:,:] = np.array([[0, 2],
                          [1, 5]])
data = data.reshape(1, 2, 2, 2)  #H, W, C
print(data.shape)
print(data[0,:,:,0])
print(data[0,:,:,1])
```

```python
eps = 0.0 #1e-5

mean = data.mean(axis=(1, 2, 3))
std  = np.sqrt(data.var(axis=(1, 2, 3)) + eps)

print(mean)
print(std)
```

```python
output = (data - mean) / std
print(data[0,0,:,:])
print(data[0,1,:,:])
```
